# Step 7: Advanced Causal Inference Analysis

**Study:** Synthetic RWE Diabetes Treatment Study  
**Comparison:** Metformin (Treatment A) vs Other diabetes medication (Treatment B)

**Methods:**
- Causal framework (exchangeability, positivity, consistency)
- Directed acyclic graph (DAG)
- Multivariable regression adjustment
- IPTW (ATE and ATT) with balance diagnostics
- Doubly robust AIPW estimation
- Bootstrap 95% confidence intervals
- DoWhy backdoor identification and EconML LinearDML
- Sensitivity and heterogeneous treatment effect analyses

**Conclusion:** Results estimate the causal effect under assumptions of no unmeasured confounding, correct model specification, and adequate overlap of treatment groups.

```python
from pathlib import Path
import sys

import pandas as pd

NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_DIR = PROJECT_ROOT / "src"
DATA_DIR = PROJECT_ROOT / "data" / "synthea_csv"
REPORT_DIR = PROJECT_ROOT / "reports"

if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from advanced_causal_inference import (
    run_advanced_causal_inference,
    CAUSAL_ASSUMPTIONS,
    CONFOUNDER_COLUMNS,
    COVARIATE_LABELS,
    TREATMENT_A,
    TREATMENT_B,
)
```

## Causal framework

| Assumption | Definition |
| --- | --- |
| Exchangeability | No unmeasured confounding given measured baseline covariates |
| Positivity | Non-zero probability of each treatment within covariate strata |
| Consistency | Observed outcome equals potential outcome under assigned treatment |

**Confounders (pre-index):** age, gender, hypertension, cardiovascular disease, obesity, hyperlipidemia, kidney disease

```python
for name, text in CAUSAL_ASSUMPTIONS.items():
    print(f"{name.title()}: {text}\n")

result = run_advanced_causal_inference(DATA_DIR, REPORT_DIR)

analysis = result.analysis_data
matched = result.matched_cohort
effects = result.effect_estimates
subgroups = result.subgroup_effects
sensitivity = result.sensitivity
balance = result.balance
bootstrap = result.bootstrap
```
